```python
import cv2
import os
import glob
import shutil

def convert_dataset_to_grayscale(base_dir, output_dir):
    # Cari semua subfolder di dalam direktori utama (train, valid, test)
    splits = ['train', 'valid', 'test']

    for split in splits:
        img_dir = os.path.join(base_dir, split, 'images')
        lbl_dir = os.path.join(base_dir, split, 'labels')

        # Cek apakah folder ada
        if not os.path.exists(img_dir) or not os.path.exists(lbl_dir):
            print(f"[-] Folder {split} tidak lengkap atau dilewati.")
            continue

        print(f"\n[+] Memproses bagian: {split.upper()}")

        # Buat folder output untuk hasil grayscale
        out_img_dir = os.path.join(output_dir, split, 'images')
        out_lbl_dir = os.path.join(output_dir, split, 'labels')
        os.makedirs(out_img_dir, exist_ok=True)
        os.makedirs(out_lbl_dir, exist_ok=True)

        # Ambil semua file gambar
        img_extensions = ['*.jpg', '*.jpeg', '*.png', '*.JPG', '*.JPEG', '*.PNG']
        image_paths = []
        for ext in img_extensions:
            image_paths.extend(glob.glob(os.path.join(img_dir, ext)))

        for img_path in image_paths:
            filename = os.path.basename(img_path)
            basename = os.path.splitext(filename)[0]
            lbl_path = os.path.join(lbl_dir, f"{basename}.txt")

            # 1. Baca Gambar asli dan ubah ke Grayscale
            img = cv2.imread(img_path)
            if img is None:
                continue

            # Mengubah warna gambar menjadi hitam putih (grayscale)
            gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

            # Simpan gambar grayscale ke folder baru
            cv2.imwrite(os.path.join(out_img_dir, filename), gray_img)

            # 2. Salin File Label tanpa modifikasi koordinat
            out_lbl_path = os.path.join(out_lbl_dir, f"{basename}.txt")
            if os.path.exists(lbl_path):
                # Menyalin file teks anotasi secara langsung karena koordinat YOLO tidak berubah
                shutil.copy(lbl_path, out_lbl_path)
            else:
                # Jika tidak ada file teks (background), buat file kosong
                open(out_lbl_path, 'w').close()

        print(f"[v] Selesai mengubah {len(image_paths)} gambar ke grayscale di {split}")

# --- KONFIGURASI JALUR FOLDER ---
# Tentukan input folder (Bisa folder dataset asli, atau folder dataset_resized sebelumnya)
INPUT_DATASET_DIR = "/content/drive/MyDrive/SIB/preprocessing/dataset-resizing-baru"
OUTPUT_DATASET_DIR = "/content/drive/MyDrive/SIB/preprocessing/dataset-grayscaling-baru"

# Jalankan fungsi konversi grayscale
convert_dataset_to_grayscale(
    base_dir=INPUT_DATASET_DIR,
    output_dir=OUTPUT_DATASET_DIR
)
print("\n[SUKSES] Semua gambar telah menjadi Grayscale dengan struktur folder baru!")
```
