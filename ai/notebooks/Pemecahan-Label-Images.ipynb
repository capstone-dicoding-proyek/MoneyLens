```python
import pandas as pd

df = pd.read_csv("/content/drive/MyDrive/SIB/dataset-fix/test/_annotations.csv")  # ganti dengan nama file kamu
df.head()
```

```python
classes = sorted(df["class"].unique())
class_map = {name: idx for idx, name in enumerate(classes)}

print(class_map)
```

```python
from collections import defaultdict

labels_dict = defaultdict(list)

for _, row in df.iterrows():
    filename = row["filename"]
    img_w = row["width"]
    img_h = row["height"]

    xmin, ymin, xmax, ymax = row["xmin"], row["ymin"], row["xmax"], row["ymax"]

    # convert ke YOLO format
    x_center = ((xmin + xmax) / 2) / img_w
    y_center = ((ymin + ymax) / 2) / img_h
    width = (xmax - xmin) / img_w
    height = (ymax - ymin) / img_h

    class_id = class_map[row["class"]]

    labels_dict[filename].append(
        f"{class_id} {x_center:.6f} {y_center:.6f} {width:.6f} {height:.6f}"
    )
```

```python
import os

output_dir = "/content/drive/MyDrive/SIB/dataset-fix/test/labels/"
os.makedirs(output_dir, exist_ok=True)

for filename, lines in labels_dict.items():
    # karena semua .jpg → aman pakai ini
    txt_name = filename.replace(".jpg", ".txt")

    with open(os.path.join(output_dir, txt_name), "w") as f:
        f.write("\n".join(lines))

print("✅ Semua label berhasil dipecah!")
```
